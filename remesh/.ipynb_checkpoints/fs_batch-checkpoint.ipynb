```python
#!/usr/bin/env python3

import os
import sys
import time
import signal
import subprocess
import threading
from dataclasses import dataclass

import numpy as np


# ============================================================
# User configuration
# ============================================================

here = "/home/nickbroussinos/Code/evolving_stokes_codex/remesh/"

verbose = "true"
resume = False
supress_outputs = 0

precond = 1
subdivisions = 12
roughness = 0
remesh_size = 0

T = 10
k = 1000
Sd = 10200
Gamma = 0
gamy = 0
chi = 1
dt = 0.0001

param = 10.0 ** np.arange(-10.5, -7, 0.5)

max_workers = 2

PRE_CLEAN_OLD_BATCH_JOBS = True

LD_PRELOAD_PATH = "/usr/lib/x86_64-linux-gnu/libstdc++.so.6"


# ============================================================
# Output filtering
# ============================================================

NOISY_PATTERNS = [
    "MathWorksServiceHost",
    "GLIBCXX_3.4.31",
    "GLIBCXX_3.4.32",
    "CXXABI_1.3.15",
    "libmwflnetwork.so",
    "libmwflcrypto.so",
    "libmwflcryptoutils.so",
    "libmwflcryptoopenssl.so",
    "libmwflurlmgrfactory.so",
]


def is_noisy(line: str) -> bool:
    return any(pattern in line for pattern in NOISY_PATTERNS)


# ============================================================
# Helpers
# ============================================================

def matlab_string(s: str) -> str:
    """
    Convert a Python string to a MATLAB single-quoted string.
    """
    return "'" + s.replace("'", "''") + "'"


def lastFrame(directory: str) -> int:
    frames = []

    for file in os.listdir(directory):
        if file.startswith("geo") and file.endswith(".mat"):
            try:
                frame = int(file.split("geo")[1].split(".")[0])
                frames.append(frame)
            except Exception:
                pass

    return max(frames) if frames else 0


def make_run_tag(Da: float) -> str:
    return (
        f"Sd_{Sd:.2e}_Da_{Da:.2e}_gamy_{gamy:+.2e}"
        .replace("+", "p")
        .replace("-", "m")
    )


def write_matlab_driver(driver_path: str, Da: float, dir_path: str, start: int) -> None:
    """
    Write a small MATLAB script for one run.

    MATLAB will execute this with:
        matlab -r "run('/path/to/_fs_batch_driver.m')"

    The script itself exits MATLAB with status 0 or 1.
    """
    code = f"""
try
    cd({matlab_string(here)});

    verbose = {verbose};
    supress_outputs = {supress_outputs};
    dir = {matlab_string(dir_path + "/")};
    start = {start};

    p = struct();
    p.roughness = {roughness};
    p.dt = {dt};
    p.T = {T};
    p.start = {start};
    p.subdivisions = {subdivisions};
    p.k = {k};
    p.remesh_size = {remesh_size};
    p.precondition_system = {precond};
    p.Da = {Da};
    p.Sd = {Sd};
    p.Gamma = {Gamma};
    p.gamy = {gamy};
    p.chi = {chi};

    fs_multi;

    exit(0);
catch ME
    disp(getReport(ME, 'extended'));
    exit(1);
end
"""

    with open(driver_path, "w") as f:
        f.write(code)


@dataclass
class Job:
    run_tag: str
    Da: float
    dir_path: str
    log_path: str
    driver_path: str


@dataclass
class RunningJob:
    job: Job
    process: subprocess.Popen
    log_file: object
    reader_thread: threading.Thread


def build_jobs() -> list[Job]:
    jobs = []

    for Da in param:
        run_tag = make_run_tag(Da)
        dir_path = os.path.join(here, "data", "fs_batch_data", run_tag)
        os.makedirs(dir_path, exist_ok=True)

        start = lastFrame(dir_path) if resume else 0

        if resume:
            print(f"[{run_tag}] resume start frame: {start}")

        log_path = os.path.join(dir_path, "matlab.log")
        driver_path = os.path.join(dir_path, "_fs_batch_driver.m")

        write_matlab_driver(
            driver_path=driver_path,
            Da=Da,
            dir_path=dir_path,
            start=start,
        )

        jobs.append(
            Job(
                run_tag=run_tag,
                Da=Da,
                dir_path=dir_path,
                log_path=log_path,
                driver_path=driver_path,
            )
        )

    return jobs


# ============================================================
# Old-job cleanup
# ============================================================

def find_old_batch_process_groups() -> set[int]:
    """
    Finds old MATLAB batch process groups from this project.

    Matches both:
      - your older command-line style containing fs_multi directly
      - this newer driver-file style containing _fs_batch_driver.m
    """
    current_pgid = os.getpgrp()

    result = subprocess.run(
        ["ps", "-u", str(os.getuid()), "-o", "pid=,ppid=,pgid=,command="],
        text=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.DEVNULL,
    )

    pgids = set()

    for line in result.stdout.splitlines():
        parts = line.strip().split(maxsplit=3)

        if len(parts) < 4:
            continue

        try:
            pid = int(parts[0])
            pgid = int(parts[2])
        except ValueError:
            continue

        cmd = parts[3]

        if pid == os.getpid():
            continue

        if pgid == current_pgid:
            continue

        is_old_inline_batch = "fs_multi" in cmd and here in cmd
        is_old_driver_batch = "_fs_batch_driver.m" in cmd and here in cmd
        is_batch_data_matlab = "matlab" in cmd.lower() and "fs_batch_data" in cmd and here in cmd

        if is_old_inline_batch or is_old_driver_batch or is_batch_data_matlab:
            pgids.add(pgid)

    return pgids


def process_group_exists(pgid: int) -> bool:
    try:
        os.killpg(pgid, 0)
        return True
    except ProcessLookupError:
        return False
    except PermissionError:
        return True


def kill_process_group(pgid: int, label: str = "", grace_seconds: float = 5.0) -> None:
    if not process_group_exists(pgid):
        return

    prefix = f"{label}: " if label else ""

    print(f"{prefix}killing process group {pgid} with SIGTERM", file=sys.stderr)

    try:
        os.killpg(pgid, signal.SIGTERM)
    except ProcessLookupError:
        return

    deadline = time.time() + grace_seconds

    while time.time() < deadline:
        if not process_group_exists(pgid):
            return
        time.sleep(0.25)

    if process_group_exists(pgid):
        print(f"{prefix}process group {pgid} survived; killing with SIGKILL", file=sys.stderr)

        try:
            os.killpg(pgid, signal.SIGKILL)
        except ProcessLookupError:
            pass


def clean_old_batch_jobs() -> None:
    pgids = find_old_batch_process_groups()

    if not pgids:
        print("No old MATLAB batch process groups found.")
        return

    print("Found old MATLAB batch process groups:")
    for pgid in sorted(pgids):
        print(f"  PGID {pgid}")

    for pgid in sorted(pgids):
        kill_process_group(pgid, label="old batch")


# ============================================================
# Launching and monitoring
# ============================================================

def stream_output(running: RunningJob) -> None:
    process = running.process
    job = running.job
    log_file = running.log_file

    assert process.stdout is not None

    for line in process.stdout:
        log_file.write(line)
        log_file.flush()

        if not is_noisy(line):
            sys.stdout.write(f"[{job.run_tag}] {line}")
            sys.stdout.flush()


def launch_job(job: Job) -> RunningJob:
    env = os.environ.copy()
    env["LD_PRELOAD"] = LD_PRELOAD_PATH

    log_file = open(job.log_path, "a", buffering=1)

    log_file.write("\n")
    log_file.write("=" * 80 + "\n")
    log_file.write(f"Starting job: {job.run_tag}\n")
    log_file.write(f"Da: {job.Da}\n")
    log_file.write(f"Directory: {job.dir_path}\n")
    log_file.write(f"Driver: {job.driver_path}\n")
    log_file.write("=" * 80 + "\n")
    log_file.flush()

    matlab_command = f"run({matlab_string(job.driver_path)})"

    process = subprocess.Popen(
        [
            "matlab",
            "-nodisplay",
            "-nosplash",
            "-nodesktop",
            "-r",
            matlab_command,
        ],
        cwd=here,
        env=env,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
        start_new_session=True,
    )

    running = RunningJob(
        job=job,
        process=process,
        log_file=log_file,
        reader_thread=None,
    )

    reader_thread = threading.Thread(target=stream_output, args=(running,), daemon=True)
    running.reader_thread = reader_thread
    reader_thread.start()

    print(f"Launched {job.run_tag} with MATLAB PID {process.pid}")

    return running


def stop_running_job(running: RunningJob) -> None:
    process = running.process

    if process.poll() is None:
        try:
            pgid = os.getpgid(process.pid)
        except ProcessLookupError:
            pgid = process.pid

        kill_process_group(pgid, label=running.job.run_tag)

    try:
        if process.stdout is not None:
            process.stdout.close()
    except Exception:
        pass

    try:
        running.reader_thread.join(timeout=2)
    except Exception:
        pass

    try:
        running.log_file.close()
    except Exception:
        pass


def close_finished_job(running: RunningJob) -> int:
    returncode = running.process.returncode

    try:
        running.reader_thread.join(timeout=2)
    except Exception:
        pass

    try:
        running.log_file.write(f"\nJob exited with code {returncode}\n")
        running.log_file.flush()
        running.log_file.close()
    except Exception:
        pass

    return returncode


def run_batch(jobs: list[Job], max_workers: int) -> None:
    pending = list(jobs)
    running: list[RunningJob] = []
    failures = 0

    try:
        while pending or running:
            while pending and len(running) < max_workers:
                job = pending.pop(0)
                running.append(launch_job(job))

            still_running = []

            for rjob in running:
                returncode = rjob.process.poll()

                if returncode is None:
                    still_running.append(rjob)
                    continue

                returncode = close_finished_job(rjob)

                if returncode == 0:
                    print(f"Finished {rjob.job.run_tag}")
                else:
                    failures += 1
                    print(
                        f"FAILED {rjob.job.run_tag} with exit code {returncode}. "
                        f"See log: {rjob.job.log_path}",
                        file=sys.stderr,
                    )

            running = still_running
            time.sleep(0.5)

    except KeyboardInterrupt:
        print("\nInterrupted. Killing active MATLAB jobs...", file=sys.stderr)

        for rjob in list(running):
            stop_running_job(rjob)

        print("Stopped active MATLAB jobs.", file=sys.stderr)
        raise SystemExit(130)

    if failures:
        raise SystemExit(failures)


# ============================================================
# Main
# ============================================================

def main() -> None:
    print(f"Project directory: {here}")
    print(f"max_workers: {max_workers}")
    print(f"resume: {resume}")
    print(f"number of jobs: {len(param)}")

    if PRE_CLEAN_OLD_BATCH_JOBS:
        clean_old_batch_jobs()

    jobs = build_jobs()

    print("Jobs to run:")
    for job in jobs:
        print(f"  {job.run_tag}")

    run_batch(jobs, max_workers=max_workers)


if __name__ == "__main__":
    main()
```
